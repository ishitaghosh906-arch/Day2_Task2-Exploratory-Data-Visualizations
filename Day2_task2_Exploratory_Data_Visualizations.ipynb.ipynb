```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Pandas Version:", pd.__version__)
print("Matplotlib Version:", plt.matplotlib.__version__)
print("Seaborn Version:", sns.__version__)

print("All libraries are ready!")
```

```python
data = {
    "Age": [22, 25, 28, 30, 32, 35, 38, 40, 45, 50],
    "Salary": [20000, 25000, 30000, 35000, 40000, 45000, 50000, 55000, 60000, 65000],
    "Experience": [1, 2, 3, 5, 6, 8, 10, 12, 15, 18],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT", "Sales", "IT", "HR", "Sales"]
}

df = pd.DataFrame(data)

df
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Salary"], kde=True)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.savefig("salary_distribution.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Feature Correlation Heatmap")

plt.savefig("correlation_heatmap.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Experience", y="Salary", hue="Department", s=100)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

plt.savefig("experience_vs_salary.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df["Salary"])

plt.title("Salary Box Plot")
plt.ylabel("Salary")

plt.savefig("salary_boxplot.png", dpi=300)
plt.show()
```

```python
import os

files = [
    "salary_distribution.png",
    "correlation_heatmap.png",
    "experience_vs_salary.png",
    "salary_boxplot.png"
]

for file in files:
    if os.path.exists(file):
        print(file, " - Saved Successfully!")
    else:
        print(file, " - Not Found!")
```
